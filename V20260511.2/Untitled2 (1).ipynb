```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 基础参数
n = [1,2,4,8,16,32,64,128,256]
pi_real = np.pi

# 2. 核心计算
pi_approx = [i * np.sin(np.pi/i) for i in n]
error = [abs(pi_real - p) for p in pi_approx]
h = [1/i for i in n]

# 3. 控制台输出
for ni, p, e in zip(n, pi_approx, error):
    print(f"n={ni:3d} | πₙ={p:.12f} | 误差={e:.2e}")

# ========== 新增：计算收敛斜率 ==========
log_e = np.log(error[-2:])
log_h = np.log(h[-2:])
slope = (log_e[1] - log_e[0]) / (log_h[1] - log_h[0])

# 4. 绘图（双对数收敛图）
plt.loglog(h, error, 'o-')
plt.grid(True, which="both")
plt.xlabel("h=1/n")
plt.ylabel("误差eₙ")

# ========== 新增：在图上标注斜率 ==========
plt.text(0.01, 1e-6, f"slope = {slope:.2f}", fontsize=12)

plt.show()
```
